```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
music = pd.read_csv(r"C:\Users\owsmi\OneDrive\Documents\GitHub\CS130-Project\music.csv")
df = pd.DataFrame(music)
popular_artists = df.loc[df['artist.hotttnesss'] > .85 , ['artist.hotttnesss', 'artist.name']]
songs2000s = df.loc[(df['song.year'] >= 2000) & (df['song.year'] < 2010), ['artist.name', 'song.year']]
```

```python
counts_pop = popular_artists.value_counts()
counts_pop.plot.bar()
plt.xlabel("Artist")
plt.ylabel("Number Of Songs")
plt.title("Number of Songs Per Artist/Group With .85 Popularity or Above")
plt.show()
```

```python
counts_year = songs2000s.value_counts("song.year")
counts_year.plot.bar()
plt.xlabel("Year")
plt.ylabel("Num Songs")
plt.title("Number of Songs Each Year in 2000s")
plt.show
```

```python
counts_genres = df.value_counts("artist.terms")
pop_genres = counts_genres[counts_genres >= 60]
pop_genres.plot.bar()
plt.xlabel("Genre")
plt.ylabel("Num Songs")
plt.title("Amount of Songs From Each Genre")
plt.show()
```
